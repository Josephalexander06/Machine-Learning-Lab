```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error,r2_score
```

```python
housing = fetch_california_housing()

data = pd.DataFrame(housing.data, columns = housing.feature_names)
data["Price"] = housing.target
# print(data.shape)   # Find row of data sets
# print(data.columns) # Find the columns
```

```python
x = data[['AveRooms']].values
y = data['Price'].values
```

```python
X_train,X_test, y_train,y_test = train_test_split(x,y, test_size=0.2, random_state=42) 
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
w,b = 0,0
learning_rate = 0.01
epochs = 1000
cost_history = []
n = len(X_train_scaled)

for i in range(epochs):
    y_pred = w * X_train_scaled.flatten() + b
    dw = (1/n) * np.sum((y_pred - y_train) * X_train_scaled.flatten())
    db = (1/n) * np.sum(y_pred - y_train)

    w , b = w - learning_rate * dw ,  b - learning_rate * db
    cost = (1/(2*n)) * np.sum((y_pred - y_train) ** 2)
    cost_history.append(cost)


    if i % 100 == 0: 
        cost = (1/(2*n)) * np.sum((y_pred - y_train) ** 2)
        print(f"Epochs {i}, Cost = {cost:.4f}")


y_pred_gd = w * X_test_scaled.flatten() + b
```

```python
print("Graident Descent")
print("--------------")
print("Weight : ", w)
print("bias :", b)
print("MSE: ", mean_squared_error(y_test,y_pred_gd))
print("R2 Score :",r2_score(y_test,y_pred_gd) )
```

```python

X_train_ne = np.c_[np.ones((len(X_train),1)),X_train]
X_test_ne = np.c_[np.ones((len(X_test),1)),X_test]

theta = np.linalg.inv(X_train_ne.T @ X_train_ne) @ X_train_ne.T @ y_train

y_pred_ne = X_test_ne @ theta
```

```python
print("Normal Equation")
print("--------------")
print("Intercept : ", theta[0])
print("Slope :", theta[1])
print("MSE: ", mean_squared_error(y_test,y_pred_ne))
print("R2 Score :",r2_score(y_test,y_pred_ne))
```

```python

plt.figure(figsize=(12, 5))

# --- Subplot 1: Gradient Descent ---
plt.subplot(1, 2, 2)
plt.scatter(y_test, y_pred_gd, color='blue', alpha=0.6)

# Dynamically find the absolute min and max for consistent bounds
limits_gd = [
    min(y_test.min(), y_pred_gd.min()),
    max(y_test.max(), y_pred_gd.max())
]
# Plot perfect diagonal identity line across the full range
plt.plot(limits_gd, limits_gd, 'r--', label='Ideal Fit')

plt.xlim(limits_gd)
plt.ylim(limits_gd)
plt.xlabel("Actual Values")
plt.ylabel("Predicted Values")
plt.title("Gradient Descent")
plt.legend()

```

```python
plt.figure(figsize=(12, 5))


# --- Subplot 2: Normal Equation ---
plt.subplot(1, 2, 2)
plt.scatter(y_test, y_pred_ne, color='green', alpha=0.6)

# Dynamically find limits for the second plot
limits_ne = [
    min(y_test.min(), y_pred_ne.min()),
    max(y_test.max(), y_pred_ne.max())
]
plt.plot(limits_ne, limits_ne, 'r--', label='Ideal Fit')

plt.xlim(limits_ne)
plt.ylim(limits_ne)
plt.xlabel("Actual Values")
plt.ylabel("Predicted Values")
plt.title("Normal Equation")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.plot(cost_history, color='blue')
plt.xlabel("Epoch")
plt.ylabel("Cost")
plt.title("Gradient Descent Cost vs Epochs")
plt.grid(True)
plt.show()
```
